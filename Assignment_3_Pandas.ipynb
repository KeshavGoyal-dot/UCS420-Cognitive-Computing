# Q6

```python
employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
})

employees.to_csv("employees.csv", index=False)

print("employees.csv created successfully.")
display(employees)
```

### a)

```python
print("Shape (rows, columns):", employees.shape)
```

### b)

```python
employees.info()
```

### c)

```python
display(employees.describe())
```

### d)

```python
print("First 5 rows:")
display(employees.head(5))

print("Last 3 rows:")
display(employees.tail(3))
```

### e)

```python
average_salary = employees["Salary"].mean()
total_bonus = employees["Bonus"].sum()
youngest_age = employees["Age"].min()
highest_rating = employees["Rating"].max()

print("i. Average salary:", average_salary)
print("ii. Total bonus paid:", total_bonus)
print("iii. Youngest employee's age:", youngest_age)
print("iv. Highest performance rating:", highest_rating)
```

### f)

```python
sorted_employees = employees.sort_values(by="Salary", ascending=False)
display(sorted_employees)
```

### g)

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(performance_category)
display(employees)
```

### h)

```python
print("Missing values in each column:")
print(employees.isnull().sum())
```

### i)



```python
employees.rename(columns={"Employee_ID": "ID"}, inplace=True)
display(employees)
```

### j)


```python
print("i. Employees with more than 5 years of experience:")
display(employees[employees["Years_of_Experience"] > 5])

print("ii. Employees belonging to the IT department:")
display(employees[employees["Department"] == "IT"])
```

### k)

```python
employees["Tax"] = employees["Salary"] * 0.10
display(employees)
```

### l)

```python
output_file = "employees_modified.csv"
employees.to_csv(output_file, index=False)

print(f"Modified DataFrame saved successfully as: {output_file}")
display(employees)
```
