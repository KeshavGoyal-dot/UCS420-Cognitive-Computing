Q3

(i)

```python
import random

random.seed(1024170318)

numbers = [random.randint(100, 900) for _ in range(100)]

print("List of 100 random numbers:")
print(numbers)
```

(ii)

```python
odd_numbers = [num for num in numbers if num % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

(iii)

```python
even_numbers = [num for num in numbers if num % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True
```

(iv)

```python
prime_numbers = [num for num in numbers if is_prime(num)]

print("\nPrime numbers:")
print(prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

(v)

```python
most_frequent = max(numbers, key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequent number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 3]


A = {digit * 7 for digit in digits}

B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

union_set = A.union(B)
print("\nUnion of A and B:", union_set)
```

(vii,viii)

```python
intersection_set = A.intersection(B)
print("\nIntersection of A and B:", intersection_set)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("\nA - B:", A_minus_B)
print("B - A:", B_minus_A)
```

(ix,x,xi)

```python
symmetric_diff = A.symmetric_difference(B)
print("\nSymmetric Difference:", symmetric_diff)

print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

X = int(input("\nEnter a value to remove from Set A: "))

A.discard(X)

print("Set A after discard():", A)
```
