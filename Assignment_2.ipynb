```python
rollNo=1024170318
my_list=[]
for digit in str(rollNo):
  x=int(digit)
  x*=10
  my_list.append(x)
print(my_list)
```

```python
my_list.append(90)
my_list
my_list.insert(3,85)
my_list
```

```python
my_list.remove(10)
my_list #10 at index 0 (first occurrence) has been removed
my_list.pop(3) #40 has been poped
```

```python
my_list.sort()
my_list
```

```python
my_list.sort(reverse=True)
my_list
```

```python
my_list[:4]
my_list
my_list[4:]
my_list
```

```python
new_list=[]
average=sum(my_list)/len(my_list)
for i in my_list:
  if i>average:
    new_list.append(i)
print(new_list)
```

```python

scores=(10,0,20,40,10,70,0,30)
highest=max(scores)
print(highest)
print(scores.index(highest))
lowest=min(scores)
print(lowest)
print(scores.count(lowest))
```

```python
import builtins
rev_scores=builtins.list(scores[::-1])
print(rev_scores)
```

```python
score=int(input("Enter a score:"))
if score in scores:
  print(scores.index(score))
else:
  print("Not presesnt")
```
