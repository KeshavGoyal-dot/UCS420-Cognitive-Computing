```python
x=np.array([1,2,3,4,5,1,2,1,1,1])
values,count=np.unique(x,return_counts=True)

most_frequent_x=values[np.argmax(count)]
indices_x=np.where(x==most_frequent_x)

print("x=",x)
print("The most frequent value in the array is:",most_frequent_x)
print("indices:",indices_x[0])



y=np.array([1,1,1,2,3,4,2,4,3,3,3])
values,count=np.unique(y,return_counts=True)

most_frequent_y=values[np.argmax(count)]
indices_y=np.where(y==most_frequent_y)

print("y=",y)
print("The most frequent value in the array is:",most_frequent_y)
print("indices:",indices_y[0])
```

Q4

```python
Keshav=np.linspace(10,100,25)

print("array:")
print(Keshav)

print("dimensions (ndim):",Keshav.ndim)

print("Shape:",Keshav.shape)

print("Total elements (size):",Keshav.size)

print("Data Type:",Keshav.dtype)

print("Total bytes consumed (nbytes):",Keshav.nbytes)

print("transposse of 1-D array:")
print(Keshav.T)

column_array=Keshav.reshape(25,1)

print("Reshaped as column:")
print(column_array)

print("Transpose of column_array:")
print(column_array.T)
```

```python
ucs420_Keshav=np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])
print("Original 2-D array:")
print(ucs420_Keshav)

print("Mean:",np.mean(ucs420_Keshav))

print("Median:",np.median(ucs420_Keshav))

print("Maximum:",np.max(ucs420_Keshav))

print("Minimum:",np.min(ucs420_Keshav))

print("Unique elements:",np.unique(ucs420_Keshav))

reshaped_ucs420_Keshav=ucs420_Keshav.reshape(4,3)
print("Reshaped array (4*3):")
print(reshaped_ucs420_Keshav)

resized_ucs420_keshav=np.resize(ucs420_Keshav,(2,3))
print("Resized array (2*3):")
print(resized_ucs420_keshav)
```
